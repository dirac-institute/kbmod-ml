```python
from hyrax import Hyrax

h = Hyrax()
```

```python
data_request = {
    "train": {
        "data": {
            "dataset_class": "kbmod_ml.data_sets.kbmod_stamps.KbmodStamps",
            "data_location": "/Users/drew/data/kbmod_ml/",
            "primary_id_field": "object_id",
            "split_fraction": 1.0,
            "dataset_config": {
                "kbmod_ml": {
                    "augment": True,
                    "true_positive_file_name": "tp_balance_debug_test.npy", #"tp_combined_5_7sigma.npy",
                    "false_positive_file_name": "fp_combined_5_7sigma_selections.npy", #"fp_combined_5_7sigma.npy",
                }
            }
        }
    },
}

h.set_config('data_request', data_request)
```

```python
ds = h.prepare()
```

```python
h.set_config('model.name', 'kbmod_ml.models.hurum_net.KBModNet')
h.set_config('train.epochs', 1)
h.set_config('optimizer', {'name': "torch.optim.AdamW"})
h.set_config('"torch.optim.AdamW"', {'lr': 3e-4, 'weight_decay': 1e-4})

h.set_config('scheduler', {'name': "torch.optim.lr_scheduler.CosineAnnealingLR"})
h.set_config('"torch.optim.lr_scheduler.CosineAnnealingLR"', {'T_max': h.config['train']['epochs'], 'eta_min': 1e-6})
```

```python
model = h.train()
```
